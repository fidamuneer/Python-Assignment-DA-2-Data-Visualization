```python
Checking Missing Values
```

```python
df.isnull().sum()
```

```python
The categorical columns (payment, pickup_zone, dropoff_zone, pickup_borough, and dropoff_borough) contain missing values.
These missing values are handled by replacing them with the mode of their respective columns.
```

```python
df["payment"] = df["payment"].fillna(df["payment"].mode()[0])
df["pickup_zone"] = df["pickup_zone"].fillna(df["pickup_zone"].mode()[0])
df["dropoff_zone"] = df["dropoff_zone"].fillna(df["dropoff_zone"].mode()[0])
df["pickup_borough"] = df["pickup_borough"].fillna(df["pickup_borough"].mode()[0])
df["dropoff_borough"] = df["dropoff_borough"].fillna(df["dropoff_borough"].mode()[0])
```

```python
df.isnull().sum()
```

```python
Removing Critical rows with missing values
The pickup and fare columns are important for our analysis because they are required for time-based and fare-related visualizations.
But in the given dataset both columns doesn't have missing values.
```

```python
df = df.dropna(subset=["pickup","fare"])
```

```python
LINE CHART
```

```python
df["pickup"] = pd.to_datetime(df["pickup"])
```

```python
Plotting every indivitual fare makes difficult to analyse. So i calculate average fare for each pickup date
```

```python
daily_fare = df.groupby(df["pickup"].dt.date)["fare"].mean()
daily_fare
```

```python
plt.figure(figsize = (12,5))
plt.plot(daily_fare.index , daily_fare.values)
plt.title("Average Fare Over Time")
plt.xlabel("Pickup Date")
plt.ylabel("Avg Fare")
plt.grid()
plt.show()
```

```python
BAR CHART
```

```python
borough_fare = df.groupby("pickup_borough")["fare"].sum()
```

```python
plt.figure(figsize = (12,5))
plt.bar(borough_fare.index , borough_fare.values)
plt.title("Total Fare by Pickup Borough")
plt.xlabel("Pickup Borough")
plt.ylabel("Total Fare")
plt.show()
```

```python
PIE CHART
```

```python
payment_count = df["payment"].value_counts()
```

```python
payment_count.plot(
    kind="pie",
    figsize = (7,7),
    autopct = "%1.1f%%",
    title = "Payment Method Distribution"
)
plt.ylabel = ("")
plt.show()
```

```python
HISTOGRAM
```

```python
plt.figure(figsize = (12,5))
plt.hist(df["distance" ], bins = 15)
plt.title("Distribution of Taxi Trip Distance")
plt.xlabel("Distance")
plt.ylabel("Frequency")
plt.show()
```

```python
BOX PLOT
```

```python
df.boxplot(
    column = "tip",
    by = "pickup_borough",
    figsize = (8,4)
)
plt.title("Tip Amount by Pickup Borough")
plt.xlabel("Pickup Borough")
plt.suptitle("")
plt.ylabel("Tip")
plt.show()
```
